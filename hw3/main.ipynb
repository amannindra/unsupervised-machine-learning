# CSE 190 HW3 – Visualization methods (Animals with Attributes)

```python
from pylab import rcParams
rcParams['figure.figsize'] = 10, 10
import numpy as np, matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
X = np.loadtxt('Data_file_awa/predicate-matrix-continuous.txt')
names = [l.split()[1].replace('+',' ') for l in open('Data_file_awa/classes.txt')]
print(X.shape, len(names))

def plot_embedding(Z, title, fname):
    plt.figure(figsize=(10,10))
    plt.scatter(Z[:,0], Z[:,1])
    print(Z[:,0])
    print(Z[:,1])
    for (x,y),n in zip(Z,names): plt.annotate(n,(x,y),fontsize=9)
    plt.title(title); plt.savefig(fname, dpi=150, bbox_inches='tight'); plt.show()
```

## (a) PCA projection to 2-d

```python
Z_pca = PCA(n_components=2).fit_transform(X)
plot_embedding(Z_pca, 'Part (a): PCA embedding', '3a_pca.png')
```

## (b) t-SNE with perplexity 5, 10, 25, 49

```python
embeds = {'PCA': Z_pca}
for p in [5,10,25,49]:
    Z = TSNE(n_components=2, perplexity=float(p), random_state=0).fit_transform(X)
    embeds[f't-SNE (perplexity={p})'] = Z
    plot_embedding(Z, f'Part (b): t-SNE, perplexity={p}', f'3b_tsne_perp{p}.png')
```

## (c) Average distortion

```python
embeds
```

```python
from scipy.spatial.distance import pdist, squareform
def avg_distortion(X, Z):
    D, Dh = squareform(pdist(X)), squareform(pdist(Z))
    c = D.mean()/Dh.mean()
    mask = ~np.eye(len(X), dtype=bool)   # diagonal is 0/0; exclude it
    r = D[mask]/(c*Dh[mask])
    return np.maximum(r, 1/r).mean()

res = {k: avg_distortion(X,Z) for k,Z in embeds.items()}
for k,v in res.items(): print(f'{k:25s} {v:.4f}')
best = min(res, key=res.get); print('Best:', best)
plt.figure(figsize=(8,5)); plt.bar(range(len(res)), list(res.values()))
plt.xticks(range(len(res)), list(res), rotation=20); plt.ylabel('average distortion')
plt.title('Part (c): average distortion per embedding')
plt.savefig('3c_distortion.png', dpi=150, bbox_inches='tight'); plt.show()
```

**Answer (c):** Average distortion: PCA 1.831, t-SNE perp 5 1.876, perp 10 1.652, perp 25 **1.591**, perp 49 1.728. The t-SNE embedding with perplexity 25 fares best (lowest distortion). t-SNE is randomized (seed fixed at 0 here), so exact values vary by run; PCA is deterministic.
