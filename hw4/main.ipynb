```python
import kagglehub
from sklearn.neighbors import NearestNeighbors
import os
from pathlib import Path

from time import perf_counter

# Download latest version
path = kagglehub.dataset_download("anmolkumar/glove-embeddings")
from sklearn.random_projection import GaussianRandomProjection

from pylab import rcParams
rcParams['figure.figsize'] = 10, 10
import numpy as np, matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
print("Path to dataset files:", path)
```

```python
l = os.listdir(path)[0]
path = Path(path) / l
path
```

```python
import numpy as np
filename = path
with open(filename) as f:
    content = f.read().splitlines()
n = len(content)
vecs = np.zeros((n,300))
print(f"Vector shape: {vecs.shape}")
words = []
index = 0
for index, rawline in enumerate(content):
    line = rawline.split()
    words.append(line[0])
    vecs[index] = np.genfromtxt(line[1:])
    index = index+1
    if index % 50000 == 0:
        print(f"Processed {index} lines")
```

```python
# print(f"Word: {words[0]}, Vector: {vecs[0]}")
```

```python
close = 6
neigh = NearestNeighbors(n_neighbors=close).fit(vecs)
word_list = ['communism', 'africa', 'happy', 'sad', 'upset', 'computer', 'cat', 'dollar']
vecs_list = []

word_to_index = {word: index for index, word in enumerate(words)}
word_index = [word_to_index[word] for word in word_list]
vecs_list = [vecs[i] for i in word_index]

# print(word_index)
# print(len(word_index))
# print(vecs_list)
# print(len(vecs_list))
# for i in word_list:
#     for j in words:
#         if i == j:
            
#             print(f"Word: {i}, Vector: {vecs[words.index(j)]}")
#             vecs_list.append(vecs[words.index(j)])
# neigh.kneighbors()
```

```python
# print(f"vecs_list: {vecs_list}")
vectors, connect = neigh.kneighbors(vecs_list)
# print(f"closest vectors: {vectors}")
# print(f"closest connect: {connect}")



dict = {}
for i in word_list:
    dict[i] = []
    
for index, i in enumerate(connect):
    word = word_list[index]
    closest_words = [words[j] for j in i]
    closest_words = closest_words[1:]
    # dict[i] = closest_words
    print(f"{word}: {closest_words}")
```

```python

start = perf_counter()
Z_pca = PCA(n_components=100).fit_transform(vecs)
end = perf_counter()

print(f"Time taken for PCA: {end - start:.4f} seconds")

neigh = NearestNeighbors(n_neighbors=close).fit(Z_pca)
word_list = ['communism', 'africa', 'happy', 'sad', 'upset', 'computer', 'cat', 'dollar']
vecs_list = []

word_to_index = {word: index for index, word in enumerate(words)}
word_index = [word_to_index[word] for word in word_list]
vecs_list = [Z_pca[i] for i in word_index]

dict = {}
for i in word_list:
    dict[i] = []
    
for index, i in enumerate(connect):
    word = word_list[index]
    closest_words = [words[j] for j in i]
    closest_words = closest_words[1:]
    # dict[i] = closest_words
    print(f"{word}: {closest_words}")
```
