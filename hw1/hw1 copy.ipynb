```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.preprocessing import normalize
```

```python
scaler = StandardScaler()

X_normalized = scaler.fit_transform(gdp)

print("Mean:", X_normalized.mean(axis=0))
print("Variance:", X_normalized.var(axis=0))
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)


labels = kmeans.fit_predict(X_normalized)

print("Cluster labels:", labels)
print("Cluster centers (normalized):")
print(kmeans.cluster_centers_)

```

```python
for i in range(5):
    cluster_points = gdp[labels == i]

    plt.scatter(
        cluster_points[:, 0],
        cluster_points[:, 1],
        label=f"Cluster {i + 1}",
        alpha=0.7
    )

centers_original = scaler.inverse_transform(
    kmeans.cluster_centers_
)

plt.scatter(
    centers_original[:, 0],
    centers_original[:, 1],
    marker="X",
    s=200,
    c="black",
    label="Cluster Centers"
)

plt.xlabel("GDP per Capita (USD)")
plt.ylabel("Infant Mortality (per 1,000)")
plt.title("K-Means Clustering of World Bank Data (k=5)")

plt.legend()
plt.grid(alpha=0.3)
plt.savefig("/Users/amannindra/cse190/9c_image1.jpg", dpi=150, bbox_inches="tight")
plt.show()
```
