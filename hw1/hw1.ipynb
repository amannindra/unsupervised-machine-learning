```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_excel("/Users/amannindra/cse190/Data_file_data2020 (2).xls")

pd.set_option("display.max_rows", None)

display(df)

countries = df.iloc[:, 0].to_list()
X = df.iloc[:, 1:].to_numpy(dtype=float)

print("Number of countries:", len(countries))
print("Shape of data:", X.shape)
```

```python
scaler = StandardScaler()

X_normalized = scaler.fit_transform(X)

print("Mean:", X_normalized.mean(axis=0))
print("Variance:", X_normalized.var(axis=0))
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X_normalized)

print("Cluster labels:", labels)
print("Cluster centers (normalized):")
print(kmeans.cluster_centers_)
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X_normalized)

print("Cluster labels:", labels)
print("Cluster centers (normalized):")
print(kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(10, 6))

for i in range(5):
    cluster_points = X[labels == i]

    plt.scatter(
        cluster_points[:, 0],
        cluster_points[:, 1],
        label=f"Cluster {i + 1}",
        alpha=0.7
    )

centers_original = scaler.inverse_transform(
    kmeans.cluster_centers_
)

plt.scatter(
    centers_original[:, 0],
    centers_original[:, 1],
    marker="X",
    s=200,
    c="black",
    label="Cluster Centers"
)

plt.xlabel("GDP per Capita (USD)")
plt.ylabel("Infant Mortality (per 1,000)")
plt.title("K-Means Clustering of World Bank Data (k=5)")

plt.legend()
plt.grid(alpha=0.3)
plt.savefig("/Users/amannindra/cse190/9c_image1.jpg", dpi=150, bbox_inches="tight")
plt.show()
```

```python
for i in range(5):

    cluster_countries = [
        countries[j]
        for j in range(len(countries))
        if labels[j] == i
    ]

    print(f"\nCluster {i + 1}:")
    print(f"Number of countries: {len(cluster_countries)}")

    for country in cluster_countries:
        print(country)
```

```python
results = pd.DataFrame({
    "Country": countries,
    "GDP per Capita": X[:, 0],
    "Infant Mortality": X[:, 1],
    "Cluster": labels + 1
})

results = results.sort_values(
    by=["Cluster", "Country"]
)

display(results)
```

**(e)** GDP per capita is in thousands of dollars (up to ~100,000+) while infant mortality is in tens per 1,000 (roughly 1-100). K-means uses Euclidean distance, so without normalizing, GDP would dominate the distance and the clusters would essentially split on GDP alone. Standardizing to mean 0 and variance 1 puts both features on the same scale so each contributes comparably to the clustering.
