```python
import pandas as pd
from sklearn.decomposition import DictionaryLearning
import matplotlib.pyplot as plt
from sklearn.datasets import make_sparse_coded_signal
```

```python
with open("Data_spokes.txt", "r", encoding="utf-8") as file:
    content = []
    for line in file:
        # print(line.strip())
        # print(line.strip().split())
        s = line.strip().split()
        # print(type(s))
        # print(type(s[0]))
        
        s[0] = float(s[0])
        s[1] = float(s[1])
        content.append(s)
    

content
```

```python
x = []
y = []
for i in content:
    x.append(i[0])
    y.append(i[1])
```

```python
plt.scatter(x, y)
plt.savefig("1a.png", dpi=300, bbox_inches="tight")
```

```python
dict_learner = DictionaryLearning(
    n_components=3, transform_algorithm='lars', transform_n_nonzero_coefs=1
)
Z = dict_learner.fit_transform(content)

dictionary = dict_learner.components_

for i, direction in enumerate(dictionary):
    plt.quiver(
        0, 0,
        direction[0], direction[1],
        angles="xy",
        scale_units="xy",
        scale=1,
        label=f"Atom {i+1}"
    )
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.savefig("1b.png", dpi=300, bbox_inches="tight")

plt.show()
```
